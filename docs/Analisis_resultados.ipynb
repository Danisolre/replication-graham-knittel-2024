3.1 Comparación de resultados

Para evaluar la reproducción del estudio se analizó el conjunto de datos generado por el repositorio original, específicamente el archivo ECF_total.csv, el cual contiene información a nivel de condado sobre la huella de carbono del empleo en Estados Unidos.

El análisis exploratorio del dataset muestra que el conjunto de datos contiene aproximadamente 3139 observaciones correspondientes a condados de Estados Unidos y 63 variables relacionadas con características demográficas, emisiones y medidas de carga de carbono. Entre las variables más relevantes se encuentran identificadores geográficos como FIPS, County y State, así como variables analíticas como POP, rho_eff_avg y burden_avg.

Los estadísticos descriptivos revelan una alta heterogeneidad entre condados en la variable burden_avg, que representa la carga promedio de carbono asociada al empleo. El valor promedio observado es aproximadamente 1.53×10⁸, mientras que el valor máximo alcanza 1.13×10¹⁰, lo que indica que algunos condados presentan niveles significativamente más altos de carga de carbono que el promedio nacional.

Esta variabilidad también se observa en la distribución poblacional. La variable POP presenta una media cercana a 104,321 habitantes, mientras que la mediana se sitúa alrededor de 25,898 habitantes, lo que sugiere una distribución altamente sesgada hacia la derecha. Este comportamiento es consistente con la estructura demográfica de Estados Unidos, donde muchos condados tienen poblaciones relativamente pequeñas y unos pocos concentran grandes cantidades de población.

Adicionalmente, se generó un histograma de la variable burden_avg, el cual muestra una distribución fuertemente asimétrica hacia la derecha. La mayoría de los condados presentan niveles relativamente bajos de carga de carbono, mientras que un número reducido de condados concentra valores extremadamente altos. Este patrón es coherente con el enfoque del estudio original, que plantea que la transición energética puede afectar de manera desigual a diferentes regiones.

3.2 Análisis de discrepancias

El análisis de discrepancias consistió en examinar posibles diferencias o inconsistencias entre los resultados obtenidos en la reproducción y lo que teóricamente se esperaría según el estudio original.

Durante la exploración del dataset se identificó una ligera discrepancia en el número de observaciones disponibles para algunas variables. Por ejemplo, mientras la variable burden_avg presenta 3139 observaciones, la variable POP contiene 3136 registros, lo que indica la presencia de valores faltantes en algunos condados.

Asimismo, se identificaron tres valores faltantes en la variable Emp, lo que sugiere que ciertos registros no contienen información completa sobre el empleo en esos condados. Este tipo de discrepancias es común en datasets construidos a partir de múltiples fuentes estadísticas y puede deberse a limitaciones en la disponibilidad o consolidación de datos.

Otra diferencia observada durante la reproducción se relaciona con la visualización geográfica de los resultados. Al generar el mapa de la huella de carbono a nivel de condado se identificaron diferencias en la representación de territorios no continentales, particularmente Alaska y Hawaii. Estas diferencias pueden depender de los métodos utilizados para unir los datos con las geometrías geográficas y del tratamiento de territorios fuera del territorio continental.

Aunque estas discrepancias no alteran el patrón general de los resultados, sí evidencian algunos desafíos prácticos asociados a la reproducción exacta de los resultados originales.
